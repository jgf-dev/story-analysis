```python
# sql_engine: bigquery
# output_variable: df
# output_mode: table
# start _sql
_sql = """
SELECT
  *
FROM
  gemini-ml-500819.gemini_stories.story_text

""" # end _sql
from google.colab.sql import bigquery as _bqsqlcell
df = _bqsqlcell.run(_sql)
df
```

```python
# prompt: get the count per unique word in the text of the `full_text_plain` column

from collections import Counter
import re
import pandas as pd

# Convert BigFrames DataFrame to a standard pandas DataFrame
pandas_df = df.to_pandas()

# Function to tokenize and count words
def count_words(text):
    if isinstance(text, str):
        words = re.findall(r'\b\w+\b', text.lower())
        return Counter(words)
    return Counter()

```

```python
# Apply the function to the 'full_text_plain' column
word_counts = pandas_df['full_text_plain'].apply(count_words)

print(len(word_counts))

total_word_counts = Counter()

```

```python
# Aggregate all word counts	otal_word_counts = Counter()
for counter in word_counts:
    total_word_counts.update(counter)

# Convert to a DataFrame for better viewing
word_counts_df = pd.DataFrame(total_word_counts.items(), columns=['word', 'count'])
word_counts_df = word_counts_df.sort_values(by='count', ascending=False).reset_index(drop=True)

word_counts_df
```
